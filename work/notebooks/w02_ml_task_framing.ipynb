## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

Answer: *As demonstrated in the dataframe above, our unit of analysis is a single pseudonymized web page (content_id). Each row contains the 90-day historical features for that specific page, and our sketched target column target_is_declining is a binary flag representing our proxy label (1 for decay, 0 for stable/growing).*

```python
import pandas as pd

# Load the starter dataset from the public repo URL
url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Sketching the target column (Proxy Label)
df['target_is_declining'] = (df['trend_direction'] == 'down').astype(int)

# Displaying the unit of analysis
display_cols = ['content_id', 'impressions_90d', 'content_age_days', 'trend_direction', 'target_is_declining']

print("Unit of Analysis: 1 Row = 1 Unique Pseudonymized Web Page\n")
print(f"Total rows (pages) loaded: {len(df)}\n")
print("Sample of the data showing our feature context and the sketched target:")
display(df[display_cols].head())
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Answer: *A fixed, hand-written rule (like if age > 180 and CTR < 2%) uses rigid thresholds. It treats all pages the same and ignores nuanced, non-linear relationships. Machine Learning beats a fixed rule here because an ML algorithm can dynamically weight the interactions between multiple features—such as how impressions, seasonality, content length, and competition all interact simultaneously—to find subtle patterns of decay that a human writing SQL rules would completely miss.*
