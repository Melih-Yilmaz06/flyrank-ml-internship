## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Answer: *Looking at the food hygiene dataset, the impact of tracking safety equipment via machine learning is striking. When no equipment is tracked or used, the average food cleanliness score drops to a dangerous 36.0%. However, when the system detects and enforces just the use of masks and gloves, the score jumps significantly to 85.1%. When all four items (hairnet, mask, sleeves, gloves) are confirmed, the cleanliness rate reaches 93.0%. These numbers prove that building a model to monitor and enforce these simple variables over the next 7 weeks can directly solve a critical health and quality problem.*

File: hygienic_dataset.csv

```python
import pandas as pd

df = pd.read_csv('hygienic_dataset.csv')

# 1. without equipment
none_used = df[(df['Bone_Kullanimi']==0) & (df['Maske_Kullanimi']==0) &
               (df['Kolluk_Kullanimi']==0) & (df['Eldiven_Kullanimi']==0)]['Temizlik_Skoru_Yuzde'].mean()

# 2. Only mask and gloves
mask_glove_only = df[(df['Maske_Kullanimi']==1) & (df['Eldiven_Kullanimi']==1)]['Temizlik_Skoru_Yuzde'].mean()

# 3. All equipment
all_used = df[(df['Bone_Kullanimi']==1) & (df['Maske_Kullanimi']==1) &
              (df['Kolluk_Kullanimi']==1) & (df['Eldiven_Kullanimi']==1)]['Temizlik_Skoru_Yuzde'].mean()

print(f"When NO equipment is used, avg score: {none_used:.1f}%")
print(f"When ONLY mask & gloves are used, avg score: {mask_glove_only:.1f}%")
print(f"When ALL equipment is used, avg score: {all_used:.1f}%")
```
