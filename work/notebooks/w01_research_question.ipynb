## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Answer: *Based on the starter dataset, we can observe the following critical points that justify the need for automated opportunity scoring:

Stat 1: Out of the 30,000 active, mature pages in our dataset, a staggering 54.21% are currently showing a downward trend (trend_direction == "down"). This indicates a massive, portfolio-wide decay issue that requires immediate triage.

Stat 2: More critically, there are 11,547 high-impact pages (with over 250 recent impressions) that are actively declining. No human content team can manually audit and evaluate 11,547 pages to decide which ones to update first.

Conclusion: This data proves the absolute necessity of a Machine Learning ranking model. Without a prioritized queue, reviewers would either guess where to start or waste time auditing pages with low recovery potential, leaving substantial organic traffic on the table.
*

```python
import pandas as pd

url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

total_pages = len(df)

target_df = df[(df['impressions_90d'] > 0) & (df['content_age_days'] >= 90)]
target_pages_count = len(target_df)

declining_pages = target_df[target_df['trend_direction'] == 'down']
declining_pct = (len(declining_pages) / target_pages_count) * 100

stale_declining = target_df[(target_df['trend_direction'] == 'down') & (target_df['impressions_90d'] >= 250)]
stale_declining_count = len(stale_declining)

print(f"Total pages in starter dataset: {total_pages}")
print(f"Target pages (active & age >= 90d): {target_pages_count}")
print(f"Percentage of target pages in decline: {declining_pct:.2f}%")
print(f"High-impact declining pages (impressions >= 250): {stale_declining_count}")
```
